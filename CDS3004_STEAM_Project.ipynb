```python
import warnings
warnings.filterwarnings("ignore")

from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, ConfusionMatrixDisplay, roc_curve
)
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

plt.rcParams["figure.dpi"] = 150
sns.set_theme(style="whitegrid")

# 1. Load Raw Dataset
df_raw = pd.read_csv("steam_games_2026.csv")
print("=== 1. RAW DATASET SUMMARY ===")
print("Raw shape:", df_raw.shape)
print("Duplicate rows:", df_raw.duplicated().sum())
print("Missing values:\n", df_raw.isna().sum())
print("Constant columns:", [c for c in df_raw.columns if df_raw[c].nunique(dropna=False) <= 1])

# 2. Data Cleaning & Scope Filtering (2024-2026 Released Games)
df = df_raw.drop_duplicates().copy()
df["Release_Date_Parsed"] = pd.to_datetime(df["Release_Date"], errors="coerce")
df["Release_Year"] = df["Release_Date_Parsed"].dt.year

df_24_26 = df[df["Release_Year"].isin([2024, 2025, 2026])].copy()
unreleased_zero_rev = (df_24_26["Total_Reviews"] == 0).sum()
df_clean = df_24_26[df_24_26["Total_Reviews"] > 0].copy()

df_clean = df_clean.drop(columns=["Steam_Deck_Status"])
df_clean["All_Tags"] = df_clean["All_Tags"].fillna("Unknown")
df_clean["Release_Month"] = df_clean["Release_Date_Parsed"].dt.month.fillna(0).astype(int)
df_clean = df_clean.reset_index(drop=True)

print("\n=== 2. DATA CLEANING LOG ===")
print(f"Step 0 - Raw rows: {len(df_raw)}")
print(f"Step 1 - Filtered to 2024-2026 rows: {len(df_24_26)}")
print(f"Step 2 - Removed unreleased/zero-review games (Total_Reviews == 0): {unreleased_zero_rev}")
print(f"Step 3 - Dropped zero-variance column 'Steam_Deck_Status' & imputed 14 missing All_Tags")
print(f"Final analysis-ready rows: {len(df_clean)}")

# 3. Feature Engineering & Target Definition
df_clean["Is_Free"] = (df_clean["Price_USD"] == 0).astype(int)
df_clean["Has_Discount"] = (df_clean["Discount_Pct"] > 0).astype(int)
df_clean["Tag_Count"] = df_clean["All_Tags"].apply(lambda x: 0 if x == "Unknown" else len(x.split(";")))
df_clean["Log_Total_Reviews"] = np.log1p(df_clean["Total_Reviews"])
df_clean["Log_Estimated_Owners"] = np.log1p(df_clean["Estimated_Owners"])
df_clean["Log_Peak_Players"] = np.log1p(df_clean["24h_Peak_Players"])

all_tags_flat = [t.strip() for tags in df_clean["All_Tags"] if tags != "Unknown" for t in tags.split(";")]
top_15_tags = [t for t, _ in Counter(all_tags_flat).most_common(15)]

tag_feature_cols = []
for t in top_15_tags:
    col = "Tag_" + t.replace(" ", "_").replace("-", "_")
    df_clean[col] = df_clean["All_Tags"].apply(lambda x: 1 if t in [i.strip() for i in x.split(";")] else 0)
    tag_feature_cols.append(col)

REVIEW_THRESHOLD = 85
df_clean["High_Review"] = (df_clean["Review_Score_Pct"] >= REVIEW_THRESHOLD).astype(int)

print("\n=== 3. TARGET DISTRIBUTION (High_Review >= 85%) ===")
print(df_clean["High_Review"].value_counts())
print("Positive class share:", round(df_clean["High_Review"].mean(), 4))

# 4. Exploratory Visualizations
fig, axes = plt.subplots(1, 3, figsize=(15, 4.2))
sns.histplot(df_clean["Price_USD"], bins=20, kde=True, ax=axes[0], color="#4C72B0")
axes[0].set_title("Price Distribution (USD)")
axes[0].set_xlabel("Price (USD)")

sns.histplot(df_clean["Review_Score_Pct"], bins=20, kde=True, ax=axes[1], color="#55A868")
axes[1].axvline(REVIEW_THRESHOLD, color="red", linestyle="--", label=f"Threshold = {REVIEW_THRESHOLD}%")
axes[1].set_title("Review Score Distribution (%)")
axes[1].set_xlabel("Positive Review Score (%)")
axes[1].legend()

top_genres = df_clean["Primary_Genre"].value_counts().head(6).index
sns.boxplot(
    data=df_clean[df_clean["Primary_Genre"].isin(top_genres)],
    x="Primary_Genre", y="Review_Score_Pct", ax=axes[2], palette="Set2"
)
axes[2].set_title("Review Score by Top 6 Genres")
axes[2].set_xlabel("Primary Genre")
axes[2].set_ylabel("Review Score (%)")
axes[2].tick_params(axis="x", rotation=25)
plt.tight_layout()
plt.savefig("fig1_eda_distributions.png", bbox_inches="tight")
plt.show()

# 5. Clustering Analysis (K-Means + Ward Hierarchical + PCA)
cluster_cols = [
    "Price_USD", "Discount_Pct", "Log_Total_Reviews",
    "Log_Estimated_Owners", "Log_Peak_Players", "Review_Score_Pct"
]
X_clust_scaled = StandardScaler().fit_transform(df_clean[cluster_cols])

k_range = range(2, 7)
sse_list, sil_list = [], []
for k in k_range:
    km_tmp = KMeans(n_clusters=k, random_state=3004, n_init=10)
    labels_tmp = km_tmp.fit_predict(X_clust_scaled)
    sse_list.append(km_tmp.inertia_)
    sil_list.append(silhouette_score(X_clust_scaled, labels_tmp))

clust_metrics_df = pd.DataFrame({"K": list(k_range), "SSE": sse_list, "Silhouette": sil_list})
print("\n=== 4. CLUSTERING K SELECTION ===")
print(clust_metrics_df.to_string(index=False))

BEST_K = 3
kmeans_final = KMeans(n_clusters=BEST_K, random_state=3004, n_init=10)
df_clean["Cluster"] = kmeans_final.fit_predict(X_clust_scaled)

ward_labels = AgglomerativeClustering(n_clusters=BEST_K, linkage="ward").fit_predict(X_clust_scaled)
ari_ward = adjusted_rand_score(df_clean["Cluster"], ward_labels)
print(f"K-Means (K={BEST_K}) vs Ward Hierarchical ARI: {ari_ward:.4f}")

cluster_profile = df_clean.groupby("Cluster")[
    ["Price_USD", "Discount_Pct", "Total_Reviews", "Estimated_Owners", "24h_Peak_Players", "Review_Score_Pct"]
].mean().round(2)
cluster_profile["Count"] = df_clean["Cluster"].value_counts().sort_index()
print("\nCluster Profiles (Mean Values):")
print(cluster_profile)

pca = PCA(n_components=2, random_state=3004)
coords = pca.fit_transform(X_clust_scaled)
centroids_2d = pca.transform(kmeans_final.cluster_centers_)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
ax1 = axes[0]
ax2 = ax1.twinx()
ax1.plot(list(k_range), sse_list, "o-", color="#4C72B0", label="SSE (Inertia)")
ax2.plot(list(k_range), sil_list, "s--", color="#DD8452", label="Silhouette Score")
ax1.set_xlabel("Number of Clusters (K)")
ax1.set_ylabel("SSE (Inertia)", color="#4C72B0")
ax2.set_ylabel("Silhouette Score", color="#DD8452")
ax1.set_title("K-Means Elbow & Silhouette Curve")

for c in range(BEST_K):
    mask = df_clean["Cluster"] == c
    axes[1].scatter(coords[mask, 0], coords[mask, 1], alpha=0.7, label=f"Cluster {c} (n={mask.sum()})")
axes[1].scatter(centroids_2d[:, 0], centroids_2d[:, 1], marker="X", s=160, c="black", label="Centroids")
axes[1].set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% var)")
axes[1].set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% var)")
axes[1].set_title(f"PCA 2D Projection of K-Means Clusters (K={BEST_K})")
axes[1].legend()
plt.tight_layout()
plt.savefig("fig2_clustering_pca.png", bbox_inches="tight")
plt.show()

# 6. Classification Modelling (Leakage-Free Pipeline + Stratified 5-Fold CV)
num_cols = [
    "Price_USD", "Discount_Pct", "Release_Month", "Tag_Count",
    "Log_Total_Reviews", "Log_Estimated_Owners", "Log_Peak_Players"
]
bin_cols = ["Is_Free", "Has_Discount"] + tag_feature_cols
cat_cols = ["Primary_Genre"]
feature_cols = num_cols + bin_cols + cat_cols

X = df_clean[feature_cols]
y = df_clean["High_Review"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=3004, stratify=y
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_cols),
        ("bin", "passthrough", bin_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
    ]
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=3004)

print("\n=== 5. DECISION TREE DEPTH TUNING (5-FOLD CV) ===")
best_depth, best_cv_acc = 2, -1.0
for d in [2, 3, 4, 5, 6, None]:
    dt_pipe = Pipeline([
        ("prep", preprocessor),
        ("clf", DecisionTreeClassifier(criterion="entropy", max_depth=d, random_state=3004))
    ])
    scores = cross_val_score(dt_pipe, X_train, y_train, cv=cv, scoring="accuracy")
    print(f"max_depth={str(d):<4} | Mean CV Acc={scores.mean():.4f} (+/- {scores.std():.4f})")
    if scores.mean() > best_cv_acc:
        best_cv_acc = scores.mean()
        best_depth = d

candidates = {
    "Dummy Baseline": DummyClassifier(strategy="most_frequent"),
    f"Decision Tree (depth={best_depth})": DecisionTreeClassifier(criterion="entropy", max_depth=best_depth, random_state=3004),
    "KNN (k=7)": KNeighborsClassifier(n_neighbors=7),
    "SVM (RBF, C=1.0)": SVC(kernel="rbf", C=1.0, probability=True, random_state=3004),
}

results_rows = []
fitted_pipes = {}
for name, clf in candidates.items():
    pipe = Pipeline([("prep", preprocessor), ("clf", clf)])
    cv_acc = cross_val_score(pipe, X_train, y_train, cv=cv, scoring="accuracy").mean()
    pipe.fit(X_train, y_train)
    fitted_pipes[name] = pipe
    y_pred = pipe.predict(X_test)
    y_prob = pipe.predict_proba(X_test)[:, 1]
    results_rows.append({
        "Model": name,
        "CV_Acc": round(cv_acc, 4),
        "Test_Acc": round(accuracy_score(y_test, y_pred), 4),
        "Precision": round(precision_score(y_test, y_pred, zero_division=0), 4),
        "Recall": round(recall_score(y_test, y_pred, zero_division=0), 4),
        "F1_Score": round(f1_score(y_test, y_pred, zero_division=0), 4),
        "ROC_AUC": round(roc_auc_score(y_test, y_prob), 4),
    })

results_df = pd.DataFrame(results_rows)
print("\n=== 6. HELD-OUT TEST EVALUATION (n=77) ===")
print(results_df.to_string(index=False))

best_model_name = results_df.iloc[1:].sort_values("F1_Score", ascending=False).iloc[0]["Model"]
best_pipe = fitted_pipes[best_model_name]

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
cm = confusion_matrix(y_test, best_pipe.predict(X_test))
ConfusionMatrixDisplay(cm, display_labels=["Standard (<85%)", "High (>=85%)"]).plot(ax=axes[0], cmap="Blues", colorbar=False)
axes[0].set_title(f"Confusion Matrix: {best_model_name}")

for name, pipe in fitted_pipes.items():
    if name == "Dummy Baseline":
        continue
    fpr, tpr, _ = roc_curve(y_test, pipe.predict_proba(X_test)[:, 1])
    auc_val = roc_auc_score(y_test, pipe.predict_proba(X_test)[:, 1])
    axes[1].plot(fpr, tpr, label=f"{name} (AUC={auc_val:.3f})")
axes[1].plot([0, 1], [0, 1], "k--", label="Random Chance")
axes[1].set_xlabel("False Positive Rate")
axes[1].set_ylabel("True Positive Rate")
axes[1].set_title("Held-Out Test ROC Curves")
axes[1].legend(loc="lower right")
plt.tight_layout()
plt.savefig("fig3_classification_eval.png", bbox_inches="tight")
plt.show()

dt_fitted = fitted_pipes[f"Decision Tree (depth={best_depth})"]
ohe_names = list(dt_fitted.named_steps["prep"].named_transformers_["cat"].get_feature_names_out(cat_cols))
all_feat_names = num_cols + bin_cols + ohe_names

plt.figure(figsize=(16, 7))
plot_tree(
    dt_fitted.named_steps["clf"],
    feature_names=all_feat_names,
    class_names=["Standard", "High_Review"],
    filled=True, rounded=True, fontsize=8, max_depth=3
)
plt.title(f"Interpretable Decision Tree Rules (Top 3 Levels Shown, Tuned Depth={best_depth})")
plt.savefig("fig4_decision_tree.png", bbox_inches="tight")
plt.show()

# 7. Association Rule Mining (Apriori on Game Tags + Threshold Sensitivity)
transactions = [
    [t.strip() for t in tags.split(";") if t.strip()]
    for tags in df_clean["All_Tags"] if tags != "Unknown"
]
te = TransactionEncoder()
df_txn = pd.DataFrame(te.fit(transactions).transform(transactions), columns=te.columns_)

print("\n=== 7. ASSOCIATION RULE THRESHOLD SENSITIVITY ===")
for supp in [0.05, 0.08, 0.10, 0.15]:
    fi = apriori(df_txn, min_support=supp, use_colnames=True)
    rl = association_rules(fi, metric="lift", min_threshold=1.20) if len(fi) > 0 else pd.DataFrame()
    rl = rl[rl["confidence"] >= 0.60] if len(rl) > 0 else rl
    print(f"min_support={supp:.2f} | Frequent Itemsets={len(fi):<4} | Rules (conf>=0.60, lift>=1.20)={len(rl)}")

freq_items = apriori(df_txn, min_support=0.08, use_colnames=True)
rules = association_rules(freq_items, metric="lift", min_threshold=1.20)
rules = rules[rules["confidence"] >= 0.60].sort_values("lift", ascending=False).copy()
rules["Rule"] = rules.apply(
    lambda r: f"{', '.join(list(r['antecedents']))} -> {', '.join(list(r['consequents']))}", axis=1
)
print("\nTop 10 Tag Association Rules (min_support=0.08, min_conf=0.60, min_lift=1.20):")
print(rules[["Rule", "support", "confidence", "lift"]].head(10).round(4).to_string(index=False))

top_rules_plot = rules.head(8).iloc[::-1]
plt.figure(figsize=(10, 4.5))
plt.barh(top_rules_plot["Rule"], top_rules_plot["lift"], color="#4C72B0")
plt.axvline(1.0, color="red", linestyle="--", label="Baseline Lift = 1.0")
plt.xlabel("Lift")
plt.title("Top 8 Tag Co-Occurrence Rules by Lift (Support >= 0.08, Confidence >= 0.60)")
plt.legend()
plt.tight_layout()
plt.savefig("fig5_association_rules.png", bbox_inches="tight")
plt.show()

# 8. Decision Making on New Hypothetical Game Examples (LO7)
new_games = pd.DataFrame([
    {
        "Game_Concept": "Indie Co-op Roguelike ($19.99)",
        "Price_USD": 19.99, "Discount_Pct": 0, "Release_Month": 5, "Tag_Count": 10,
        "Log_Total_Reviews": np.log1p(15000), "Log_Estimated_Owners": np.log1p(450000),
        "Log_Peak_Players": np.log1p(2500), "Is_Free": 0, "Has_Discount": 0,
        "Primary_Genre": "Indie",
        "Tags_List": ["Singleplayer", "Co-op", "Roguelike", "Indie", "Strategy"]
    },
    {
        "Game_Concept": "Free-to-Play Competitive Shooter ($0)",
        "Price_USD": 0.0, "Discount_Pct": 0, "Release_Month": 10, "Tag_Count": 10,
        "Log_Total_Reviews": np.log1p(50000), "Log_Estimated_Owners": np.log1p(1500000),
        "Log_Peak_Players": np.log1p(12000), "Is_Free": 1, "Has_Discount": 0,
        "Primary_Genre": "Action",
        "Tags_List": ["Action", "Multiplayer", "FPS", "Shooter", "PvP"]
    },
    {
        "Game_Concept": "Full-Price AAA Action RPG ($69.99)",
        "Price_USD": 69.99, "Discount_Pct": 0, "Release_Month": 9, "Tag_Count": 10,
        "Log_Total_Reviews": np.log1p(8000), "Log_Estimated_Owners": np.log1p(250000),
        "Log_Peak_Players": np.log1p(4000), "Is_Free": 0, "Has_Discount": 0,
        "Primary_Genre": "RPG",
        "Tags_List": ["Singleplayer", "RPG", "Open World", "Story Rich", "Adventure"]
    }
])

for t in top_15_tags:
    col = "Tag_" + t.replace(" ", "_").replace("-", "_")
    new_games[col] = new_games["Tags_List"].apply(lambda lst: 1 if t in lst else 0)

X_new = new_games[feature_cols]
new_games["Predicted_High_Review"] = best_pipe.predict(X_new)
new_games["High_Review_Prob"] = best_pipe.predict_proba(X_new)[:, 1].round(4)

print("\n=== 8. DECISION MAKING ON NEW GAME EXAMPLES ===")
print(new_games[["Game_Concept", "Price_USD", "Primary_Genre", "Predicted_High_Review", "High_Review_Prob"]].to_string(index=False))
```
